```python
import numpy as np
import matplotlib.pyplot as plt

# Generate synthetic data (300 samples, 10 features as typical for PCA demo)
def generate_data(n_samples=300, n_features=10, n_clusters=3):
    np.random.seed(42)
    X = np.random.randn(n_samples, n_features)

    # Add correlated structure for PCA to reveal
    for i in range(n_clusters):
        start = i * (n_samples // n_clusters)
        end = (i + 1) * (n_samples // n_clusters)
        X[start:end, :3] += 2 * np.sin(np.linspace(0, 4*np.pi, end-start))[:, np.newaxis]
    return X

# Task 1: NumPy-only PCA implementation
class NumpyPCA:
    def __init__(self, n_components=2):
        self.n_components = n_components
        self.components_ = None
        self.mean_ = None
        self.explained_variance_ = None
        self.explained_variance_ratio_ = None
        self.std_ = None # Store std for inverse_transform

    def fit(self, X):
        # Step 1: Standardize the data (zero mean, unit variance)
        self.mean_ = np.mean(X, axis=0)
        self.std_ = np.std(X, axis=0) # Store std for inverse_transform
        X_centered = X - self.mean_
        X_std = X_centered / self.std_

        # Step 2: Compute covariance matrix
        cov_matrix = np.cov(X_std.T)

        # Step 3: Eigen decomposition (NumPy only)
        eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)

        # Step 4: Sort by descending eigenvalues
        idx = np.argsort(eigenvalues)[::-1]
        self.components_ = eigenvectors[:, idx[:self.n_components]]
        self.explained_variance_ = eigenvalues[idx[:self.n_components]]

        # Add self.explained_variance_ratio_ for direct comparison (moved inside fit)
        total_var = np.sum(eigenvalues) # Sum all eigenvalues for total variance
        self.explained_variance_ratio_ = self.explained_variance_ / total_var

        # Step 5: Project data onto top PCs
        self.X_transformed_ = X_std @ self.components_
        return self

    def transform(self, X):
        X_centered = X - self.mean_
        X_std = X_centered / self.std_
        return X_std @ self.components_

    def inverse_transform(self, X_transformed):
        # Reconstruct scaled data
        X_reconstructed_scaled = X_transformed @ self.components_.T
        # Unstandardize (multiply by std and add mean)
        X_reconstructed = X_reconstructed_scaled * self.std_ + self.mean_
        return X_reconstructed

# Task 2: Comparison with sklearn PCA
from sklearn.decomposition import PCA as SklearnPCA
from sklearn.preprocessing import StandardScaler

# Generate data
X = generate_data()

# NumPy PCA
np_pca = NumpyPCA(n_components=2)
np_pca.fit(X)
X_np = np_pca.transform(X)

# Sklearn PCA (for comparison)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
sk_pca = SklearnPCA(n_components=2)
X_sk = sk_pca.fit_transform(X_scaled)

print("NumPy PCA explained variance ratio:", np_pca.explained_variance_ / np.sum(np_pca.explained_variance_)) # Still using initial explained_variance_
print("Sklearn PCA explained variance ratio:", sk_pca.explained_variance_ratio_)
print(f"PC1 correlation (NumPy vs Sklearn): {np.corrcoef(X_np[:,0], X_sk[:,0])[0,1]:.4f}")

# Task 3: Mathematical explanation (your own words for submission)
"""
Principal Component Analysis (PCA) dimensionality reduction works through 5 key steps:

1. *Standardization*: Subtract mean and divide by standard deviation from each feature → Z-score normalization
2. *Covariance Matrix*: Compute Σ = (1/n)XᵗX where X contains standardized features → captures feature correlations
3. *Eigen-decomposition*: Solve Σv = λv → eigenvectors (v) are principal components, eigenvalues (λ) are variance explained
4. *Component Selection*: Sort by λ (descending) → keep top-k eigenvectors forming projection matrix W
5. *Projection*: Y = XW → transform original data into k-dimensional space maximizing variance retention

Geometric intuition: PCA finds orthogonal directions of maximum variance, rotating data to align with these axes.
"""

# Task 4: Visualization + Validation
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Original data (first 2 dims for visualization)
axes[0,0].scatter(X[:, 0], X[:, 1], alpha=0.6)
axes[0,0].set_title('Original Data (2D projection)')

# NumPy PCA results
axes[0,1].scatter(X_np[:, 0], X_np[:, 1], c=np.arange(len(X_np)), cmap='viridis', alpha=0.6)
axes[0,1].set_title('NumPy PCA (PC1 vs PC2)')

# Sklearn PCA results
axes[1,0].scatter(X_sk[:, 0], X_sk[:, 1], c=np.arange(len(X_sk)), cmap='viridis', alpha=0.6)
axes[1,0].set_title('Sklearn PCA (PC1 vs PC2)')

# Scree plot (variance explained)
n_components = 5
np_pca_full = NumpyPCA(n_components=n_components)
np_pca_full.fit(X)
sk_pca_full = SklearnPCA(n_components=n_components)
sk_pca_full.fit(X_scaled)

# Fix scree plot (your NumPy is raw eigenvalues; match sklearn's ratio/cumsum)
axes[1,1].plot(range(1, n_components+1), np_pca_full.explained_variance_ratio_.cumsum(), 'o-', label='NumPy PCA')
axes[1,1].plot(range(1, n_components+1), np.cumsum(sk_pca_full.explained_variance_ratio_), 's-',
               label='Sklearn PCA')
axes[1,1].set_xlabel('Principal Components')
axes[1,1].set_ylabel('Cumulative Variance Explained')
axes[1,1].legend()
axes[1,1].set_title('Scree Plot Comparison')

plt.tight_layout()
plt.show()

# Final validation: Check reconstruction error
# Use the new inverse_transform method
mse_np = np.mean((X - np_pca.inverse_transform(X_np))**2)
print(f"NumPy PCA reconstruction MSE: {mse_np:.6f}")
```
