```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs

# 1. Generate synthetic data
X, _ = make_blobs(n_samples=500, centers=5, n_features=2, random_state=42)

# 2. K-Means from scratch
def kmeans(X, k, max_iter=100):
    n, d = X.shape
    centroids = X[np.random.choice(n, k, replace=False)]
    for _ in range(max_iter):
        distances = np.linalg.norm(X[:, None] - centroids, axis=2)
        labels = np.argmin(distances, axis=1)
        new_centroids = np.array([X[labels == i].mean(axis=0) for i in range(k)])
        if np.allclose(centroids, new_centroids):
            break
        centroids = new_centroids
    return labels, centroids

# 3. Silhouette Score from scratch
def silhouette_score(X, labels):
    n = len(X)
    score = 0
    for i in range(n):
        same = X[labels == labels[i]]
        other = X[labels != labels[i]]
        a = np.mean(np.linalg.norm(same - X[i], axis=1))
        b = np.min([np.mean(np.linalg.norm(X[labels == l] - X[i], axis=1))
                    for l in set(labels) if l != labels[i]])
        score += (b - a) / max(a, b)
    return score / n

# 4. Multiple runs
best_score = -1
for _ in range(10):
    labels, centroids = kmeans(X, 5)
    score = silhouette_score(X, labels)
    if score > best_score:
        best_score = score
        best_labels = labels
        best_centroids = centroids

print("Best Silhouette Score:", best_score)

# 5. Plot
plt.scatter(X[:,0], X[:,1], c=best_labels)
plt.scatter(best_centroids[:,0], best_centroids[:,1], marker='X')
plt.show()
```
